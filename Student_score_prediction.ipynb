```python
#text data replacing
x['Parental_Involvement']= x['Parental_Involvement'].replace({'Low': 1, 'Medium': 2, 'High': 3})
x['Access_to_Resources']= x['Access_to_Resources'].replace({'Low': 0, 'Medium': 1, 'High': 2})
x['Extracurricular_Activities']= x['Extracurricular_Activities'].replace({'No': 0, 'Yes': 1})
x['Motivation_Level']= x['Motivation_Level'].replace({'Low': 0, 'Medium': 1, 'High': 2})
x['Internet_Access']= x['Internet_Access'].replace({'No': 0, 'Yes': 1})
x['Family_Income']= x['Family_Income'].replace({'Low': 0, 'Medium': 1, 'High': 2})
x['Teacher_Quality']= x['Teacher_Quality'].replace({'Low': 0, 'Medium': 1, 'High': 2})
x['Peer_Influence']= x['Peer_Influence'].replace({'Negative': 0 , 'Neutral': 1, 'Positive': 2})
x['Gender']=x['Gender'].replace({'Male': 1, 'Female': 0 })
x['Learning_Disabilities']= x['Learning_Disabilities'].replace({'No': 0, 'Yes': 1})






```

```python
#One hot encoding
cols= ['School_Type', 'Parental_Education_Level', 'Distance_from_Home']
x= pd.get_dummies(x, columns=cols, drop_first=True)
```

# Scaling

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler_x= MinMaxScaler(feature_range=(0,1))
scaler_y= MinMaxScaler(feature_range=(0,1))
```

```python
x= scaler_x.fit_transform(x)
y= scaler_y.fit_transform(np.array(y).reshape(-1,1))
```

# Split the data into training and test sets

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
len(y_train)
```

# Linear Regression

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
y_pred = model.predict(x_test)

plt.scatter(y_test, y_pred)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel ('Actual Score')
plt.ylabel ('Predicted Score')
plt.title ('Regression Actual vs Prediction')
plt.show()
```
